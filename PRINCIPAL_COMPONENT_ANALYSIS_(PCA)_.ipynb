```python
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[
    ~((df[numerical_columns] < lower) |
      (df[numerical_columns] > upper)).any(axis=1)
]

print("Shape after removing outliers:", df.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

plt.figure(figsize=(18, 10))

for i, col in enumerate(numerical_columns, 1):
    plt.subplot(4, 4, i)
    sns.boxplot(x=df[col])
    plt.title(f'After Removing Outliers - {col}')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
sns.histplot(df['Alcohol'], kde=True)

plt.subplot(1,3,2)
sns.histplot(df['Malic_Acid'], kde=True)

plt.subplot(1,3,3)
sns.histplot(df['Color_Intensity'], kde=True)

plt.tight_layout()

plt.show()
```

```python
X = df.drop('Customer_Segment', axis=1)

y = df['Customer_Segment']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled.shape)
```

```python
print(X_scaled[:5])
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA()
```

```python
pca.fit(X_scaled)
```

```python
print(pca.explained_variance_ratio_)
```

```python
import numpy as np

print(np.cumsum(pca.explained_variance_ratio_))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o')

plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")

plt.title("PCA Explained Variance")

plt.grid(True)

plt.show()
```

```python
pca_2 = PCA(n_components=2)
```

```python
X_pca_2 = pca_2.fit_transform(X_scaled)
```

```python
print(X_pca_2.shape)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    X_pca_2[:,0],
    X_pca_2[:,1],
    alpha=0.6
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title("PCA Projection (2D)")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    X_pca_2[:,0],
    X_pca_2[:,1],
    c=y,
    cmap='viridis',
    s=60
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title("Wine Dataset PCA")

plt.colorbar(label="Customer Segment")

plt.show()
```

```python
print("Explained Variance:")

print(pca_2.explained_variance_ratio_)
```

```python
print("Total Explained Variance:")

print(np.sum(pca_2.explained_variance_ratio_))
```

```python
print("Explained Variance Ratio:")
print(pca_2.explained_variance_ratio_)
```

```python
print("Total Explained Variance:")
print(pca_2.explained_variance_ratio_.sum())
```

```python
from sklearn.metrics import mean_squared_error

X_inverse = pca_2.inverse_transform(X_pca_2)

mse = mean_squared_error(X_scaled, X_inverse)

print("Reconstruction Error:", mse)
```

```python
from sklearn.decomposition import PCA

for i in range(1, X_scaled.shape[1] + 1):

    pca = PCA(n_components=i)

    pca.fit(X_scaled)

    print(i, "Components :", pca.explained_variance_ratio_.sum())
```

```python
pca_best = PCA(n_components=0.95)

X_pca_best = pca_best.fit_transform(X_scaled)

print("Number of Components:", pca_best.n_components_)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA(n_components=2))
])

X_pipeline = pipeline.fit_transform(X)

print(X_pipeline.shape)
```
